```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv("../.env")
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)
```

```python
with open("bloodwork.txt","r") as f:
    blood_report=f.read()
    print(blood_report[:200])
```

```python
extraction_prompt=f""" 
you are a medical data extraction assistant

from the blood report below extract all the test values and classify each 
one as HIGH,LOW,or NORMAL
based on the reference ranges provided in the report

Format your response as:
-Test name:value| Status:HIGH/NORMAL/LOW | Reference:range
Blood Report
{blood_report}
"""

extraction_response=llm.invoke(extraction_prompt)
extracted_value=extraction_response.text

print("=== STAGE 1: EXTRACTED VALUES===")
print(extracted_value)
```

```python
diet_prompt=f"""
You are a clinical nutritionist specializing in Indian Dietary Habits.

Based on teh bloodwork analysis below wirte:
1.a short health summary in 2-3 lines explaining the patients condition in simple language.
2.A short practical indian diet plan having only two sections
     (1)Food to avoid
      (2)Food to eat more of.
Do no include any other section in diet plan.
Blood work analysis:
{extracted_value}
"""
diet_response=llm.invoke(diet_prompt)
print("===STAGE 2: HEALTH SUMMARY & DIET PLAN===")
print(diet_response.text)
```
